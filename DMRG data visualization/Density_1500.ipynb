```python
import getpass
from pathlib import Path
from Physical_Quantity.Physical_Data_Record import DensityRecord
from Physical_Quantity.Density_Process import DensityProcessor
```

```python
#----------------------------------------------------------------------------------------------------------------------
#---------------------########### UltraDMRG计算密度平均值<n>的数据提取与作图 ################--------------------------

### 一、输入数据
## 1.1 基本参数
N = 200  # 格点数
a = 1.0  # honeycomb晶格常数

## 1.2 文件夹路径（需要修改的部分）
version = "49"   
base_dir = Path(f'Data_N{N}/t_J_3/d1_50/1500_measure_phys_qty/measure_{version}')

## 1.3 读取文件并提取数据
data_n = (base_dir / "nf.json").read_text(encoding='utf-8')


### 二、通过面向对象程序进行数据处理
## 2.1 将输入的数据封装成DensityRecord对象列表
n_UltraDMRG = DensityRecord.extract_data(data_n, N, a)  

## 2.2 打印n的前5个数据
print("\n----------------------------- honeycomb晶格点的电子密度分布数据 -----------------------------------")
for record in n_UltraDMRG[:5]:
    print(record)
print(f"... 还有{len(n_UltraDMRG)-5}个点")


### 三、数据可视化
## 3.1 电子密度的一维折线图可视化
DensityProcessor.Line_Chart_Visualization(n_UltraDMRG)
```
