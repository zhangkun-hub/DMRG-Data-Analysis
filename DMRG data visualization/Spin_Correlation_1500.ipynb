```python
import getpass
from pathlib import Path
from Physical_Quantity.Physical_Data_Record import CorrelationRecord
from Physical_Quantity.Correlation_Process import CorrelationProcessor
```

```python
#------------------------------------------------------------------------------------------------------------------------------
#---------------------########### A子格下的自旋-自旋关联函数<S_i·S_j>数据提取与作图 ################--------------------------

### 一、输入数据
## 1.1 基本参数
N = 200  # 格点数
a = 1.0  # honeycomb晶格常数

## 1.2 文件夹路径（需要修改的部分）
version = "49"   
base_dir = Path(f'Data_N{N}/t_J_3/d1_100/1500_measure_phys_qty/measure_{version}')

## 1.3 文件名称字典
file_mapping = {
    'SuSd': f'sm{version}sp.json',
    'SdSu': f'sp{version}sm.json',
    'SzSz': f'sz{version}sz.json',
}

## 1.4 读取文件并提取数据
# A键方向数据(对应蓝色bond)
data_SuSd = (base_dir / file_mapping['SuSd']).read_text(encoding='utf-8')
data_SdSu = (base_dir / file_mapping['SdSu']).read_text(encoding='utf-8')
data_SzSz = (base_dir / file_mapping['SzSz']).read_text(encoding='utf-8')


### 二、生成A子格序列 
source_point = int(N / 4 / 4) * 4 + 1         # 源点
field_point_max = int(N / 4 / 4 * 3) * 4 - 3  # 最大场点
A_sequence = range(source_point, field_point_max + 1, 4) 


### 三、通过面向对象程序进行数据处理
## 3.1 将输入的数据封装成CorrelationRecord对象列表
SzSz = CorrelationRecord.extract_data(data_SzSz, A_sequence, N, a)
SuSd = CorrelationRecord.extract_data(data_SdSu, A_sequence, N, a)
SdSu = CorrelationRecord.extract_data(data_SuSd, A_sequence, N, a)

## 3.2 通过CorrelationProcessor类中的Combined_spin_correlation方法得到<S_i·S_j>的CorrelationRecord对象列表
SiSj_A_Ultra = CorrelationProcessor.Combined_spin_correlation(SzSz, SuSd, SdSu)

## 3.3 打印SiSj的前5个数据
print("\n----------------------------- 自旋关联函数数据 -----------------------------------")
for record in SiSj_A_Ultra[:5]:
    print(record)
print(f"... 还有{len(SiSj_A_Ultra)-5}个点")


### 四、数据可视化
CorrelationProcessor.Spin_Corr_Visualization(SiSj_A_Ultra)
```
